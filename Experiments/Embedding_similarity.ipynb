```python
class HiddenLayer(nn.Module):
    def __init__(self, base_model):
        super().__init__()
        self.features = nn.Sequential(*list(base_model.premodel.children())[:-1])  # all layers before fc

    def forward(self, x):
        x = self.features(x)
        return x.view(x.size(0), -1)
```
